# 01. Průzkumová analýza dat (EDA) korpusu Toki Pona
Tento notebook slouží k analýze textového korpusu připraveného pro diplomovou práci:
- Rozdělení délek vět (znaky, slova)
- Zastoupení jednotlivých zdrojů (Tatoeba, Wikipesija, Lipu Sewi, Poki Lapo)
- Frekvenční analýza slovníku pu / ku
- Srovnání komprese jednotlivých tokenizérů (Char, BPE, Word)

```python
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

data_path = Path('../../data/processed/train.jsonl')
records = []
with open(data_path, 'r', encoding='utf-8') as f:
    for line in f:
        records.append(json.loads(line))

df = pd.DataFrame(records)
print(f'Celkový počet trénovacích vět: {len(df):,}')
df.head()
```

```python
plt.figure(figsize=(8, 4))
df['source'].value_counts().plot(kind='bar', color='#2a9d8f')
plt.title('Rozložení zdrojů v trénovací sadě Toki Pona')
plt.xlabel('Zdroj')
plt.ylabel('Počet vět')
plt.tight_layout()
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
df['char_len'].hist(bins=40, ax=ax1, color='#264653')
ax1.set_title('Histogram délek vět (ve znacích)')
ax1.set_xlabel('Počet znaků')

df['word_len'].hist(bins=30, ax=ax2, color='#e76f51')
ax2.set_title('Histogram délek vět (ve slovech)')
ax2.set_xlabel('Počet slov')
plt.tight_layout()
plt.show()
```
