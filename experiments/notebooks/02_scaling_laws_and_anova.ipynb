# 02. Statistická analýza škálovacích zákonů a ANOVA
Tento notebook slouží pro detailní statistické modelování výsledků experimentů:
- Načtení konsolidovaných výsledků `results/metrics/results.csv`
- Odhad parametrů Kaplan/Chinchilla škálovacích zákonů
- Faktorová ANOVA a výpočet velikostí účinku $\eta^2$
- Lineární smíšené modely (LMEM) se zohledněním náhodného efektu seedu
- Export grafů ve vysokém rozlišení pro LaTeX diplomové práce

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
import statsmodels.formula.api as smf
from pathlib import Path

results_path = Path('../../results/metrics/results.csv')
df = pd.read_csv(results_path)
df.head()
```

```python
# ANOVA a dekompozice rozptylu
formula = 'val_bpc ~ C(data_fraction) + C(model_size) + C(tokenizer) + C(data_fraction):C(tokenizer)'
model = smf.ols(formula, data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
total_ss = anova_table['sum_sq'].sum()
anova_table['eta_sq_percent'] = (anova_table['sum_sq'] / total_ss) * 100
anova_table
```
